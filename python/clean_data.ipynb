```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:/Users/Kadhilat/Desktop/retail-sales-analysis/data/superstore_raw.csv")

```

```python

#Inspect the Data
df.head()
```

```python
# Provide Summary of a Data Frame
print(df.info())
```

```python
#Print Missing Values
print("Missing Values")
print(df.isnull().sum())
```

```python
#check for dublicates
# Check for Duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Convert the date formats
for column in ["Order Date", "Ship Date"]:
    df[column] = pd.to_datetime(
        df[column],
        format = "%d/%m/%Y",
        errors= "coerce"
    )

```

```python
for column in ["Sales", "Profit", "Quantity", "Discount"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
#Check for missing values again
print("Missing values after cleaning")
print(df.isnull().sum())
```

```python
# Check unresolved values before exporting
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Dates saved in ISO format and export
df.to_csv (r"C:/Users/Kadhilat/Desktop/retail-sales-analysis/data/superstore_clean.csv", index=False) 
```
